```python
import re
import requests
from bs4 import BeautifulSoup
import os
from datetime import datetime

CSS_SELECTOR = 'div.g-user-content[data-qa="vacancy-description"]'

def fetch_html(url: str) -> str:
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
    }
    resp = requests.get(url, headers=headers, timeout=20)
    resp.raise_for_status()
    return resp.text

def extract_text(html: str) -> list:
    soup = BeautifulSoup(html, "lxml")
    nodes = soup.select(CSS_SELECTOR)
    texts = []
    for node in nodes:
        text = node.get_text(separator="\n", strip=True)
        text = re.sub(r"\n{3,}", "\n\n", text)
        texts.append(text)
    return texts

def save_to_file(texts: list, url: str, filename: str) -> str:
    os.makedirs("results", exist_ok=True)
    filepath = os.path.join("results", filename)
    
    with open(filepath, 'w', encoding='utf-8') as f:
        # f.write(f"URL: {url}\n")
        # f.write(f"Date: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n")
        
        for i, text in enumerate(texts, 1):
            if len(texts) > 1:
                f.write(f"=== Блок {i} ===\n")
            f.write(text)
            if i < len(texts):
                f.write("\n\n")
    
    return filepath

url = "https://rabota.by/vacancy/127131925?from=applicant_recommended&hhtmFrom=main"  

html = fetch_html(url)
texts = extract_text(html)

if texts:
    saved_path = save_to_file(texts, url, "vacancy.txt")
    print(f"Saved: {saved_path}")
    
    for i, t in enumerate(texts, 1):
        if len(texts) > 1:
            print(f"=== Block {i} ===")
        print(t)
else:
    print(f"Tag is not found: {CSS_SELECTOR}")
```

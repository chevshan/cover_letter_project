```python
import requests     
import json          
import time    
import os       
import pandas as pd 
```

```python
url = "https://api.hh.ru/vacancies/126886105"

request = requests.get(url)
data = request.content.decode()
request.close()

parsed_json_hh = json.loads(data)
```

```python
html_string = parsed_json_hh['description']
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(html_string, 'html.parser')
description = soup.get_text()
print(description)
```

```python
print(json.dumps(parsed_json_hh, indent=3, ensure_ascii=False))
```
